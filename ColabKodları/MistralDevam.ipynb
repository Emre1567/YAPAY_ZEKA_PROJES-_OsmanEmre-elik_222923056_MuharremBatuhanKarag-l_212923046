```python
import os
import torch
from datasets import load_dataset
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    TrainingArguments,
)
from trl import SFTTrainer

# --- KRİTİK AYAR: WANDB SORUSUNU KAPAT ---
os.environ["WANDB_DISABLED"] = "true"

# 1. AYARLAR
model_name = "mistralai/Mistral-7B-v0.1"
new_model_name = "mistral-7b-turkce-duzeltici"

#1 LoRA Yapılandırması - GÜNCELLENDİ
peft_config = LoraConfig(
    r=16,
    lora_alpha=32, # Daha kararlı bir öğrenme için artırıldı
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=[
        "q_proj", "k_proj", "v_proj", "o_proj",
        "gate_proj", "up_proj", "down_proj", # MLP katmanları eklendi: Mantık kurması için kritik!
    ]
)

#2 A100 için Optimize Edilmiş Kuantizasyon (bf16 kullanıyoruz)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16, # A100 desteği için bf16
)

# 2. MODEL VE TOKENIZER YÜKLEME
print("Model yükleniyor...")
model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto"
)
model.config.use_cache = False
model.config.pretraining_tp = 1

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right"

model = prepare_model_for_kbit_training(model)
model = get_peft_model(model, peft_config)

# 3. VERİ SETİ VE FORMATLAMA
print("Veri seti yükleniyor...")

# Veri setini yükle
dataset = load_dataset("json", data_files="ham_veri_seti_target.json", split="train")

#3
def formatting_prompts_func(example):
    output_texts = []

    # example içindeki satır sayısı kadar dön
    for i in range(len(example['raw_model_output'])):
        # Yeni vizyona uygun "Zenginleştirme/Paraphrase" talimatı
        text = (
            f"### Instruction:\n"
            f"Sen profesyonel bir Türkçe paraphrase modülüsün. Girdi metnindeki anlamı kesinlikle koruyarak, "
            f"daha zengin, akıcı ve kurallı bir Türkçe ile yeniden ifade et. Metne asla yeni bilgi ekleme.\n\n"
            f"### Input:\n{example['raw_model_output'][i]}\n\n"
            f"### Response:\n{example['corrected_output'][i]}{tokenizer.eos_token}"
        )
        # Oluşturulan metni listeye ekle
        output_texts.append(text)

    return output_texts

# Not: tokenizer objesinin bu fonksiyondan önce tanımlanmış olması gerekir.

# 4. EĞİTİM PARAMETRELERİ (A100 ODAKLI) - GÜNCELLENDİ
training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=3, # 3 Epoch tam kararında
    per_device_train_batch_size=4,
    gradient_accumulation_steps=4, # Adımları biriktirerek daha sağlam güncellemeler yapar
    optim="paged_adamw_32bit",
    save_steps=100,
    logging_steps=10,
    learning_rate=1e-4, # Biraz daha yavaş ve sağlam öğrenme (2e-4'ten 1e-4'e çekildi)
    weight_decay=0.01,  # Ezberlemeyi (overfitting) önleyen fren mekanizması
    fp16=False,
    bf16=True,
    max_grad_norm=0.3,
    max_steps=-1,
    warmup_ratio=0.1,   # Başlangıçta modele ısınma süresi tanır
    group_by_length=True,
    lr_scheduler_type="cosine", # SABİT YERİNE COSINE: Sonlara doğru hassaslaşır
    report_to=None
)

# 5. TRAINER BAŞLATMA
print("Eğitim başlıyor...")
trainer = SFTTrainer(
    model=model,
    train_dataset=dataset,
    peft_config=peft_config,
    max_seq_length=1024,
    dataset_text_field=None,
    formatting_func=formatting_prompts_func,
    tokenizer=tokenizer,
    args=training_args,
    packing=False,
)

trainer.train()

# 6. KAYDETME
print("Eğitim tamamlandı. Model kaydediliyor...")
trainer.save_model(new_model_name)
print(f"Model '{new_model_name}' adıyla başarıyla kaydedildi.")
```

```python
# outputu kısa tut
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import torch

# --- AYARLAR ---
base_model_name = "mistralai/Mistral-7B-v0.1"
new_model_name = "mistral-7b-turkce-duzeltici"

# --- MODEL YÜKLEME ---
print("Model ve Tokenizer yükleniyor...")
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16, # DÜZELTİLDİ: A100 için bf16
)

base_model = AutoModelForCausalLM.from_pretrained(
    base_model_name,
    quantization_config=bnb_config,
    device_map="auto"
)

model_to_test = PeftModel.from_pretrained(base_model, new_model_name)
tokenizer = AutoTokenizer.from_pretrained(base_model_name)
tokenizer.pad_token = tokenizer.eos_token

# --- TEST GİRDİSİ ---
test_input = "Sistemde güvenlik açığı fark edilince tüm kullanıcıların erişimini durdurduk."

# --- PROMPT (BOŞLUKLARA VE SATIRLARA DİKKAT) ---
# F-string içindeki boşluklar modelin performansı için kritiktir.
prompt = f"""### Instruction:
Sen profesyonel bir Türkçe paraphrase modülüsün. Girdi metnindeki anlamı kesinlikle koruyarak, daha zengin, akıcı ve kurallı bir Türkçe ile yeniden ifade et. Metne asla yeni bilgi ekleme.

### Input:
{test_input}

### Response:
"""

inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

# Modeli test moduna almayı unutma
model_to_test.eval()

# --- GENERATION AYARLARI (EN İYİ SONUÇ İÇİN) ---
outputs = model_to_test.generate(
    **inputs,
    max_new_tokens=100,
    do_sample=True,          # Yaratıcılık için True
    temperature=0.3,         # Mantık ve zenginlik dengesi (0.1 yerine 0.3)
    top_p=0.9,
    repetition_penalty=1.1,  # Kelime uydurmayı önlemek için (1.5 yerine 1.1)
    eos_token_id=tokenizer.eos_token_id,
    pad_token_id=tokenizer.pad_token_id
)

print("\n" + "="*30)
print("GİRDİ:", test_input)
print("-" * 30)
print("YENİ ZENGİN ÇIKTI:")

result = tokenizer.decode(outputs[0], skip_special_tokens=True)

# ÇIKTI TEMİZLEME (Hayati Kısım):
if "### Response:" in result:
    # Response'tan sonrasını al ve eğer model tekrar ### yazarsa orada kes
    response_part = result.split("### Response:")[1].strip()
    clean_output = response_part.split("###")[0].strip()
    print(clean_output)
else:
    print(result)
print("="*30)
```

```python
# --- ONE-SHOT PROMPT TEKNİĞİ ---
# Modele önce bir örnek gösteriyoruz, sonra kendi sorumuzu soruyoruz.

test_input = "Veritabanı baglanti hatasi verince sistem crash oldu ve loglar silindi."

prompt = f"""### Instruction:
Sen profesyonel bir Türkçe redaktörüsün. Aşağıdaki bozuk metinleri dilbilgisi kurallarına uygun, akıcı ve resmi bir Türkçe ile yeniden yaz. Asla yorum yapma, sadece düzeltilmiş metni ver.

### Örnek 1:
Input:
Dosya acilmiyor hata verdi galiba bozuk.
Response:
Dosya açılmıyor, muhtemelen bozuk olduğu için hata verdi.

### Gerçek Soru:
Input:
{test_input}

### Response:
"""

inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

outputs = model_to_test.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.1,        # Çok düşük sıcaklık (Yaratıcılığı öldür, tutarlılığı artır)
    repetition_penalty=1.2,
    top_p=0.9,              # Sadece en mantıklı kelimeleri seç
    pad_token_id=tokenizer.eos_token_id
)

print("\n" + "="*30)
print("GİRDİ:", test_input)
print("-" * 30)
result = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Çıktıyı temizleyip alalım
if "### Gerçek Soru:" in result:
    # Promptun sonrasını al
    part = result.split("### Response:\n")[-1]
    print("SONUÇ:", part.strip())
else:
    print(result)
print("="*30)
```

```python
# --- 3-SHOT (ÜÇ ÖRNEKLİ) PROMPT ---
# Modele tam olarak ne istediğimizi 3 farklı örnekle gösteriyoruz.

test_input = "Veritabanı baglanti hatasi verince sistem crash oldu ve loglar silindi."

# Prompt'u bir "Dönüştürücü" gibi kurguluyoruz.
prompt = f"""Aşağıdaki görevde, verilen bozuk Türkçe cümleleri sadece dilbilgisi kurallarına göre düzelt. Asla cümlenin anlamını değiştirme, yeni bilgi ekleme veya yorum yapma.

Örnek 1:
Input: Dosya acilmiyor hata verdi galiba bozuk.
Output: Dosya açılmıyor; muhtemelen bozuk olduğu için hata verdi.

Örnek 2:
Input: yarin okula gelcem mi bilmiyom
Output: Yarın okula gelecek miyim, bilmiyorum.

Örnek 3:
Input: pc cok ısındı kapandı
Output: Bilgisayar çok ısındı ve kapandı.

Görev:
Input: {test_input}
Output:"""

inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

outputs = model_to_test.generate(
    **inputs,
    max_new_tokens=50,      # Cevabı kısa tutmaya zorluyoruz
    do_sample=True,
    temperature=0.01,       # Sıfıra yakın sıcaklık (Yaratıcılık İSTEMİYORUZ)
    repetition_penalty=1.1,
    pad_token_id=tokenizer.eos_token_id
)

print("\n" + "="*30)
print("GİRDİ:", test_input)
print("-" * 30)
result = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Sadece son cevabı ayıklama
if "Output:" in result:
    # Son "Output:" ifadesinden sonrasını al
    final_answer = result.split("Output:")[-1].strip()
    print("SONUÇ:", final_answer)
else:
    print("HAM SONUÇ:", result)
print("="*30)
```

```python
import gradio as gr
from transformers import pipeline

# 1. Pipeline Kurulumu (En Güvenli Yöntem)
# Bu yöntem, karmaşık ayarları (temperature, stop token) otomatik yönetir.
print("Pipeline hazırlanıyor...")
generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=128, # Cevap uzunluğu
    do_sample=True,
    temperature=0.4,    # 0.01 yerine 0.4 (Model nefes alsın)
    top_p=0.90,
    repetition_penalty=1.1
)

def duzeltici_bot(input_text):
    # Prompt'u basitleştiriyoruz. Karmaşık örnekler (Few-Shot)
    # şu anki modelin kafasını karıştırıyor olabilir.
    # Mistral'in doğal formatını ([INST]) kullanıyoruz.

    prompt = f"""[INST] Sen bir Türkçe redaktörüsün. Aşağıdaki bozuk metni dilbilgisi kurallarına uygun olarak düzelt:

    {input_text} [/INST]"""

    # Üretim
    output = generator(prompt)
    generated_text = output[0]['generated_text']

    # Sadece cevabı ayıklama kısmı
    if "[/INST]" in generated_text:
        return generated_text.split("[/INST]")[1].strip()
    return generated_text

# --- ARAYÜZ TASARIMI ---
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("""
    # 🇹🇷 Türkçe AI Redaktör (Final)
    **Mistral-7B** modeli kullanılarak hazırlanmıştır.
    """)

    with gr.Row():
        with gr.Column():
            input_box = gr.Textbox(
                label="Bozuk Metin",
                placeholder="Örn: Veritabanı baglanti hatasi verince sistem crash oldu...",
                lines=3
            )
            submit_btn = gr.Button("✨ Düzelt", variant="primary")

            gr.Examples(
                examples=[
                    ["Veritabanı baglanti hatasi verince sistem crash oldu."],
                    ["yarin aksam bize gelcekmisin?"],
                    ["Dosya yüklenirken hata olustu tekrar deneyin."]
                ],
                inputs=input_box
            )

        with gr.Column():
            output_box = gr.Textbox(label="Düzeltilmiş Metin", lines=3, interactive=False)

    submit_btn.click(fn=duzeltici_bot, inputs=input_box, outputs=output_box)

print("Arayüz başlatılıyor... Linke tıkla!")
demo.launch(share=True, debug=True)
```

```python
# --- KURTARMA KODU (Relax Mod) ---
def duzeltici_bot_final(input_text):

    # Prompt'u biraz sadeleştirelim, modelin kafası karışmasın
    prompt = f"""### Instruction:
Sen bir Türkçe redaktörüsün. Aşağıdaki bozuk metni, anlamını değiştirmeden sadece dilbilgisi kurallarına uygun hale getir.

### Örnek:
Input: server patladı veriler uçtu
Output: Sunucu çöktü ve veriler silindi.

### Görev:
Input: {input_text}
Output:"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")

    outputs = model.generate(
        **inputs,
        max_new_tokens=100,
        do_sample=True,

        # --- AYARLARI GEVŞETİYORUZ ---
        temperature=0.3,        # 0.01 çok düşüktü, 0.3 idealdir.
        top_p=0.9,              # Kelime havuzunu biraz genişletelim.
        repetition_penalty=1.1, # 1.2 çok sertti, 1.1 yapalım.

        pad_token_id=tokenizer.eos_token_id,
        eos_token_id=tokenizer.eos_token_id
    )

    result = tokenizer.decode(outputs[0], skip_special_tokens=True)

    if "Output:" in result:
        return result.split("Output:")[-1].strip()
    return result

# --- TEST ET ---
girdi = "Veritabanı baglanti hatasi verince sistem crash oldu ve loglar silindi."
print("GİRDİ:", girdi)
print("SONUÇ:", duzeltici_bot_final(girdi))
```
